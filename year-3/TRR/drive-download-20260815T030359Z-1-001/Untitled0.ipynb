```python
def count_arrangements(n, a, b, c, mem):
    if (a < 0 or b < 0 or c < 0):
        return 0
    if (n == 0):
        return 1
    if (mem[n][a][b][c] != -1):
        return mem[n][a][b][c]

    result = 0
    if (a > 0):
        result += count_arrangements(n - 1, a - 1, b, c, mem)
    if (b > 0):
        result += count_arrangements(n - 1, a, b - 1, c, mem)
    if (c > 0):
        result += count_arrangements(n - 1, a, b, c - 1, mem)

    mem[n][a][b][c] = result
    return result

# Initialize memoization table
n, a, b, c = 9, 3, 3, 3
mem = [[[[ -1 for _ in range(c + 1)] for _ in range(b + 1)] for _ in range(a + 1)] for _ in range(n + 1)]

# Calculate the number of valid arrangements
result = count_arrangements(n, a, b, c, mem)
print(result)
```
